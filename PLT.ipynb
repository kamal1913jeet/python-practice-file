```python
import matplotlib.pyplot as plt
a = ["mon" ,"tues","wed","thurs","fri","sat"]
b = [20,13,26,32,28,16]

plt.style.use("ggplot")

plt.plot(a,b ,color = "red" , linestyle="--" , linewidth = 1
         , marker="o" , label = "2025 sales" )
plt.title("Sales of this week")
plt.xlabel("days of the week")
plt.ylabel("sales in a day")
plt.grid(color = "gray" , linestyle= ":" ,linewidth= 1)
plt.legend(loc ="upper left")


# range of x axis and y axis
plt.xlim(0,8)
plt.ylim(0,50)


# to name the points on x side like mon tue wed etc from 1,2,3,4etc
# plt.xticks([list of old index] ,[list of new names])      


plt.show()
```

```python
import matplotlib.pyplot as plt
product = ["handbag" , "goggle" , "bracelets" , "belt"]
sales = [1200 , 2100, 850 , 1800]

plt.bar(product , sales , color = "red" , label = "sales in last week" , width = 0.5)

plt.title("Sales 2025")
plt.xlabel("PRODUCTS")
plt.ylabel("SALES IN A WEEK")

plt.legend(loc ="upper left")
plt.show()
```

```python
import matplotlib.pyplot as plt
product = ["handbag" , "goggle" , "bracelets" , "belt"]
sales = [1200 , 2100, 850 , 1800]
plt.pie(sales , labels = product , autopct =" %1.1f%%" )
plt.title("sales shown in pie chart")
plt.show()
```

UNIFORM DATA DISTRIBUTION


```python
import numpy as np
import matplotlib.pyplot as plt

# Generate uniform data
uniform_data = np.random.uniform(low=0, high=10, size=1000)

# Plot
plt.hist(uniform_data, bins=60)
plt.title("Uniform Distribution")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate normal data
normal_data = np.random.normal(loc=50, scale=5, size=1000)

# Plot
plt.hist(normal_data, bins=20, density = True)
plt.title("Normal Distribution")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()
```
