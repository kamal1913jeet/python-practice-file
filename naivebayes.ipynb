```python
import pandas as pd
df=pd.read_csv("Titanic-Dataset.csv")
df.head()
```

```python
df.drop(['PassengerId',	'Name','Parch','Ticket','Cabin','Embarked'], axis='columns',inplace=True)
df.head()
```

```python
target= df.Survived
inputs = df.drop('Survived',axis='columns') #droping target variable nd want to separte this
inputs
```

```python
dumie= pd.get_dummies(inputs.Sex)
dumie.head(5)
```

```python
#now we will concat these dummy variable into our dataset
inputs=pd.concat([inputs,dumie],axis='columns')
inputs.head(5)
```

```python
inputs.drop('Sex',axis='columns',inplace=True)
inputs.head(10)
```

```python
#to check if their is any NAN value in any column
#it tells us that the age column has  nainputs.columns[inputs.isna().any()] n value
```

```python
inputs.Age[:10] #printed first 10 values in order to check the nan really exusts in our code
#we can handle thses nan values by taking the mean of the values nd then using fiilna option in order to fill those nan place
```

```python
inputs.Age=inputs.Age.fillna(inputs.Age.mean())
inputs.head(10)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test= train_test_split(inputs,target,test_size=0.2)
```

```python
from sklearn.naive_bayes import GaussianNB
model = GaussianNB()
```

```python
model.fit(X_train,Y_train)
```

```python
model.score(X_test,Y_test)
```

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
df = pd.read_csv("dectree.csv")
df.head()
```

```python
le_company = LabelEncoder()
le_job = LabelEncoder()
le_degree = LabelEncoder()

df['company_n'] = le_company.fit_transform(df['company'])
df['job_n'] = le_job.fit_transform(df['job'])
df['degree_n'] = le_degree.fit_transform(df['degree'])
df=df.drop(["company","job","degree"], axis=1)
```

```python
X = df[['company_n','job_n','degree_n']]
y = df['salary_more_then_100k']
df.head()
```

```python

from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(X, y, test_size=0.2)


x_train.shape , x_test.shape
```

```python
from sklearn.naive_bayes import CategoricalNB
model= CategoricalNB()

model.fit(x_train,y_train)
```

```python
print(model.predict(0,1,0))
print(y_test)
print(model.score(x_test,y_test))
```
