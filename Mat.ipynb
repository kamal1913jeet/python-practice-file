```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.array([1,2,3,4,5])
y = np.array([10,20,30,40,50])
plt.plot(x,y)
plt.title("Data")
plt.xlabel("X Data")
plt.ylabel("Y Data")
plt.show()
```

```python
# Bar chart
students = ['A','B','C','D']
marks = [80,90,100,110]

plt.bar(students,marks)
plt.title("Bar Chart")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()
```

```python
# Pie chart
sizes = [40,20,30,10]
xyz = ['python','Java','C++','JS']

plt.pie(sizes,labels=xyz,autopct='%1.0f%%')
plt.title("Pie Chart")
plt.show()
```

```python
# Scatter plot:
x = [1,2,3,4,5]
y = [20,10,10,30,12]

plt.scatter(x,y,color="red",marker = "*",s=25)
plt.title("Scatter Plot")
plt.ylabel('Y Axis')
plt.xlabel("X Axis")
plt.show()
```

```python
data = [10,20,30,40,10,20,30,40,40,50]
plt.hist(data)
plt.title("Histogram")
plt.show()
```

```python
# Area Chart
x = [1,2,3,4]
y = [3,4,1,2]

plt.fill_between(x,y)
plt.title("Area Plot")
plt.show()
```

```python
# Box Plot
data = [12,15,14,19,20,22]
plt.boxplot(data)
plt.title("Box Plot")
plt.show()
```

```python
data = [12,10,15,19,20,22]
plt.violinplot(data)
plt.title("Violin PLOT")
plt.show()
```

```python
# step plot
x = [1,2,3,4]
y = [10,20,30,40]
plt.step(x,y)
plt.show()
```

```python
# stem plot
plt.stem(x,y)
plt.show()
```

```python
# HorIZONTAL bAR PLOT
x = ['A','B','C']
y = [10,20,10]
plt.barh(x,y)
plt.show()
```

```python
# subplots
x = [1,2,3,3]
y = [10,12,9,7]
plt.subplot(1,2,1)
plt.plot(x,y)

plt.subplot(1,2,2)
plt.plot(x,y)
plt.show()
```
