```python
import pandas as pd 
from sklearn.datasets import load_digits
digit=load_digits()
```

```python
dir(digit)
```

```python
%matplotlib inline
import matplotlib.pyplot as plt 
plt.gray()
for i in range(4):
    plt.matshow(digit.images[i])
```

```python
df=pd.DataFrame(digit.data)
df
```

```python
df['target']=digit.target #target is showing us the truth like here we can we the no. from 0 to 4 that depicts the digits above i.e from 0-4
df
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(df.drop(['target'],axis='columns'),digit.target,test_size=0.2)
```

```python
len(X_train)
```

```python
from sklearn.ensemble import RandomForestClassifier
model=RandomForestClassifier(n_estimators=40) #n_estimators parameter in the model specifies the number of decision trees in the forest
model.fit(X_train,Y_train)
```

```python
model.score(X_test,Y_test)
```

CONFUSION MATRIX----> A confusion matrix is a tool used in machine learning to evaluate the performance of a classification model. It is a matrix that summarizes the number of correct and incorrect predictions made by the model on a set of test data. The matrix is represented as a table with rows and columns representing the predicted and actual classes, respectively. The diagonal elements of the matrix represent the number of correct predictions, while the off-diagonal elements represent the number of incorrect predictions. The confusion matrix provides a deeper layer of insight into the model's performance and can be used to calculate various metrics such as accuracy, precision, recall, and F1 score These metrics can help in understanding the model's strengths and weaknesses and can be used to fine-tune the model for better performance


```python
y_predicted=model.predict(X_test)
```

```python
from sklearn.metrics import confusion_matrix
cm=confusion_matrix(Y_test,y_predicted)
cm
```

```python
%matplotlib inline
import matplotlib.pyplot as plt
import seaborn as sn
plt.figure(figsize=(10,7))
sn.heatmap(cm, annot=True)
plt.xlabel('Predicted')
plt.ylabel('Truth')
```
