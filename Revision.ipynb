```python
import pandas as pd
import numpy as np

# 1. CREATE SERIES FROM LIST

list1 = [10, 20, 30, 40, 50]

s1 = pd.Series(list1)
print(s1)
```

```python
# 2. CREATE SERIES FROM NUMPY ARRAY

array1 = np.array([1, 2, 3, 4, 5])

s2 = pd.Series(array1)
print(s2)
```

```python
# 3. CREATE SERIES FROM DICTIONARY

dictionary1 = {
    'A': 100,
    'B': 200,
    'C': 300,
    'D': 400
}

s3 = pd.Series(dictionary1)
print(s3)
```

```python
# 4. CUSTOM INDEXING

s4 = pd.Series(
    [85, 90, 78, 92],
    index=['Rahul', 'Amit', 'Neha', 'Priya']
)
print(s4)
print("\n")
# Access by label
print(s4['Neha'])
print("\n")
# # Multiple labels
print(s4[['Rahul', 'Priya']])
```

```python
# 5. CUSTOM OPERATIONS - SQAURE, MULTIPLY, DIVIDE 
array1 = np.array([0, 2, 3, 4, 5])

s2 = pd.Series(array1)
print("\nSeries from NumPy Array:")
print(s2)

print("\n SQAURE")
print(s2**2)
print("\nMULTIPLY")
print(s2*10)
print("\nDIVIDE")
print(s2/10)
```

```python
# 6. CREATE DATAFRAME USING MULTIPLE SERIES

name = pd.Series(['Rahul', 'Amit', 'Neha', 'Priya'])
age = pd.Series([21, 22, 20, 23])
marks = pd.Series([85, 90, 78, 92])

df1 = pd.DataFrame({
    'Name': name,
    'Age': age,
    'Marks': marks
})
print(df1)
```

```python
# 7. CONVERT SERIES TO DICTIONARY

student_marks = pd.Series(
    [85, 90, 78],
    index=['Rahul', 'Amit', 'Neha']
)

marks_dict = student_marks.to_dict()

print(marks_dict)
```

```python
# 8. SERIES -> DICTIONARY --> DATAFRAME

df2 = pd.DataFrame(
    list(marks_dict.items()),
    columns=['Student', 'Marks']
)
print(df2)
```

```python
# 9. DICTIONARY TO DATAFRAME

student_data = {
    'Name': ['Rahul', 'Amit', 'Neha', 'Priya'],
    'Age': [21, 22, 20, 23],
    'Marks': [85, 90, 78, 92]
}

df3 = pd.DataFrame(student_data)
print(df3)
```

```python
# 10. CUSTOM INDEX IN DATAFRAME

df4 = pd.DataFrame(
    student_data,
    index=['S1', 'S2', 'S3', 'S4']
)
print(df4)

print("\nAccess row using custom index")
print("\n")
print(df4.loc['S2'])
```
